### Question 1 : une estimation de la distribution de Y

Les colonnes correspondent aux vraies valeurs, on fait donc la somme de chaque colonne.

$g_1$ : $A = 44 + 5 + 1 = 50$, $B = 0 + 62 + 8 = 70$, $C = 0 + 1 + 54 = 55$

$g_2$ : $A = 44 + 2 + 4 = 50$, $B = 4 + 64 + 2 = 70$, $C = 5 + 3 + 47 = 55$

$|\mathcal{D}'| = 50 + 70 + 55 = 175$

$P(Y = A) = \frac{50}{175} = \frac{2}{7} \approx 0{,}29 \qquad P(Y = B) = \frac{70}{175} = \frac{2}{5} = 0{,}40 \qquad P(Y = C) = \frac{55}{175} = \frac{11}{35} \approx 0{,}31$

Pour vérifier : $\frac{2}{7} + \frac{2}{5} + \frac{11}{35} = \frac{10 + 14 + 11}{35} = 1$

### Question 2 : Les contrôles à faire sur D et D'

- Les deux matrices sont évaluées sur le même $\mathcal{D}'$ : elles doivent avoir le même total (175) et les mêmes sommes de colonnes (50, 70, 55). C'est le cas.
- $\mathcal{D}$ et $\mathcal{D}'$ doivent être disjoints, sinon l'évaluation est trop optimiste.
- Les classes de $\mathcal{D}'$ doivent toutes être présentes dans $\mathcal{D}$.
- $\mathcal{D}$ et $\mathcal{D}'$ doivent suivre la même distribution (proportions de classes proches).

### Question 3 : Accuracy de chaque modèle de D'

Les bonnes prédictions sont sur la diagonale ($p = t$).

$\text{Acc}(g_1) = \frac{44 + 62 + 54}{175} = \frac{160}{175} \approx 0{,}914$

$\text{Acc}(g_2) = \frac{44 + 64 + 47}{175} = \frac{155}{175} \approx 0{,}886$

### Question 4 : meilleur modèle entre $g_1$ et $g_2$ selon le risque empirique

$L_b(p,t) = 0$ si $p = t$ et $1$ sinon. Chaque case contribue $n(p,t) \times L_b(p,t)$ :

$\hat{R}(g) = \frac{1}{|\mathcal{D}'|} \sum_{p} \sum_{t} n(p,t) \, L_b(p,t) = 1 - \text{Acc}(g)$

$\hat{R}(g_1) = \frac{5 + 1 + 8 + 1}{175} = \frac{15}{175} \approx 0{,}086$

$\hat{R}(g_2) = \frac{4 + 5 + 2 + 3 + 4 + 2}{175} = \frac{20}{175} \approx 0{,}114$

$\hat{R}(g_1) < \hat{R}(g_2)$, donc on choisit $g_1$.

### Question 5 : meilleur modèle entre $g_1$ et $g_2$ selon le risque

Non, on ne peut pas en être sûr. Le risque calculé en question 4 est un risque empirique : c'est seulement une estimation du vrai risque $R(g) = \mathbb{E}[L_b(g(X), Y)]$, qu'on ne peut pas calculer car on ne connaît pas la vraie distribution de $(X, Y)$.

Ici, l'écart entre les deux modèles est faible (5 erreurs sur 175, soit environ 3 %) et $\mathcal{D}'$ ne contient que 175 exemples. Avec un autre ensemble de test, le classement pourrait s'inverser.

Pour être plus confiant, il faudrait un ensemble de test plus grand, ou répéter l'évaluation sur plusieurs ensembles (validation croisée), ou encore calculer des intervalles de confiance sur les risques.

### Question 6 : Calcul du risque empirique de chaque modèle

Même formule qu'à la question 4, avec $\ell_2$ à la place de $L_b$ :

$\hat{R}(g) = \frac{1}{|\mathcal{D}'|} \sum_{p} \sum_{t} n(p,t) \, \ell_2(p,t)$

La diagonale de $\ell_2$ vaut 0, on ne garde que les cases hors diagonale.

$\hat{R}(g_1) = \frac{5 \times 1 + 1 \times 2 + 8 \times 1 + 1 \times 1}{175} = \frac{16}{175} \approx 0{,}091$

$\hat{R}(g_2) = \frac{2 \times 1 + 4 \times 2 + 5 \times 1 + 3 \times 1 + 4 \times 2 + 2 \times 1}{175} = \frac{28}{175} \approx 0{,}16$

$\hat{R}(g_1) < \hat{R}(g_2)$, donc avec $\ell_2$ on choisit encore $g_1$.

```python
import numpy as np 

def risk_emp(M):
    accuracy = (np.trace(M))/np.sum(M)
    return 1-accuracy

def best_mod(M,N):
    if M.shape != N.shape :
        print("ERROR: les deux modèles n'ont pas la même taille.")
        return 

    if not np.array_equal(np.sum(M, axis=0), np.sum(N, axis=0)):
        print("ERROR : les deux matrices n'ont pas les mêmes sommes de colonnes (pas évaluées sur le même D').")
        return
        
    r1 = risk_emp(M)
    r2 = risk_emp(N)
    
    if r1 < r2 :
        print("Le meilleur modèle est le modèle 1")
        best_model = M
    elif r2 < r1:
        print("Le meilleur modèle est le modèle 2")
        best_model = N
    else:
        print("Les deux modèles ont le même risque donc aucun n'est meilleur que l'autre.")
        best_model = None
        
    return best_model   

g1 = np.array([[44, 0, 0], [5, 62, 1], [1, 8, 54]])
g2 = np.array([[44, 4, 5], [2, 64, 3], [4, 2, 47]])
best_mod(g1, g2)
```
